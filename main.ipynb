```python
n=int(input("enter a positive integer nubmer:"))
sum=0
numbers_divisible3=0
for i in range(1,n+1):
    if i%3==0 and i%5==0:print(" 3 5 ")
    elif i%3==0:print("3")
    elif i%5==0:print("5")
    else:print(n)
    if i%2==0:sum+=i
    if i%3==0:numbers_divisible3+=1
print(f"sum:{sum}   the number of numbers divisible by 3:{numbers_divisible3}")
```
